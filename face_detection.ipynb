```python
import urllib.request
import os
import cv2
```

```python
cap = cv2.VideoCapture(0)
```

```python
#print(cv2.data.haarcascades)
```

```python

xml_url = (
    "https://raw.githubusercontent.com/opencv/opencv/"
    "4.x/data/haarcascades/"
    "haarcascade_frontalface_default.xml"
)

xml_path = os.path.abspath("haarcascade_frontalface_default.xml")

urllib.request.urlretrieve(xml_url, xml_path)

face_cascade = cv2.CascadeClassifier(xml_path)

print("Classifier loaded:", not face_cascade.empty())
```

```python
def detect_face(img):
    cordinates = face_cascade.detectMultiScale(img)
    return cordinates
    
```

```python
while True:
    ret, frame = cap.read()

    # Check whether a frame was successfully captured
    if not ret or frame is None:
        print("Failed to read frame. Check the camera or video file.")
        break

    coordinates = detect_face(cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY))
    
    for x,y,w,h in coordinates:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),1)
        
    cv2.imshow("window", frame)

    if cv2.waitKey(1) & 0xFF == ord('x'):
        break

```

```python
cap.release()
cv2.destroyAllWindows()
```
